```python

class Bank:
    def __init__ (self, ac_number, name):
        self.ac_number = ac_number
        self.name = name
        self.balance = 10000
        print("Account Detail:\nAccount number:",self.ac_number,"\nName of customer:",self.name,"\nBalance:",self.balance)
        self.options=int(input("Enter the options :{1. deposit, 2.withdrawls, 3.check_balance}: "))

        self.work()

    def deposit(self,amount):
        self.balance+=amount
        return self.balance

    def withdrawls(self):
        if self.balance <= 1000:
            print("minimum balance should be 1000")
        else:
            amount=int(input("enter the amount to withdrawls: "))
            if self.balance < amount:
                print("insufficient balance")
            else :
                self.balance-=amount
                return self.balance

    def check_balance(self):
        return "Current balance is:" + str(self.balance)

    def work(self):
        if self.options==1:
            amount=int(input("enter the amount to deposit: "))
            print(self.deposit(amount))
        elif self.options==2:
            print(self.withdrawls())
        elif self.options==3:
            print(self.check_balance())
        else:
            print("invalid option")

cust=Bank(12345,"vaibhav")
```
